# Day 2 EDA

```python
new_df = df.copy()
```

```python
new_df.isnull().sum()

# isn()
new_df.isna().sum()
```

```python
# drop()
new_df.drop(columns=["school"])
new_df


# drop index
new_df.drop(index=0)
```

```python
# dropna()
new_df.dropna()

# subset
new_df.dropna(subset=["school"])
```

```python
# filnna()
new_df["school"].mode()
```

```python
new_df["school"]=new_df["school"].fillna(new_df["school"].mode()[0])
```

```python
# numerical
new_df["age"].median()
```

```python
new_df["age"].fillna(new_df["age"].median()).astype(int)
```

```python
# dupliactes
# False, first,last
new_df[new_df.duplicated(keep="last")]
```

```python
new_df=new_df.drop_duplicates()
```

```python
new_df.head()
```

```python
new_df["at_risk"]=(new_df["G3"]<10).astype(int)
```

```python
new_df.head()
```

```python
features = [
    "studytime", "failures", "schoolsup", "famsup", "activities",
    "higher", "internet", "famrel", "health", "absences"
]

X=new_df[features].copy()
```

```python
X
```

```python
Y=new_df["at_risk"]
```

```python
Y
```

```python
class_balance= Y.value_counts()
```

```python
model_data=X.copy()
```
